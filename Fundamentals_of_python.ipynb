````python
import sys

# 1. *objects  -> "2026-07-24", "ERROR", "Database connection failed"
# 2. sep       -> " | "
# 3. end       -> " [LOGGED]\n"
# 4. file      -> sys.stdout
# 5. flush     -> True

print("2026-07-24", "ERROR", "Database connection failed", sep=" | ", end=" [LOGGED]\n", file=sys.stdout, flush=True)
````

**Deep Dive: How Each Parameter Behaved**
* **objects (The Arguments: "2026-07-24", "ERROR", "Database connection failed"):** The asterisk * means this parameter accepts an unlimited number of arguments. In this example, we passed three separate string arguments, and Python grouped them together to be printed.
* **sep=" | " (The Separator):** By default, Python injects a blank space (" ") between multiple objects. By changing sep to " | ", Python joined our three strings using a pipe character instead of a space.
* **end=" [LOGGED]\n" (The Line Ending):** Normally, print() automatically appends a newline character (\n) to the very end of your text so the next print statement starts below it. We overrode this to append a custom status tag  [LOGGED] right before jumping to the new line.
* **file=sys.stdout (The Destination Output Stream):** This tells Python where to send the text. By default, it sends text to sys.stdout (your terminal screen). However, you could switch this target to an open text file object (like file=open("log.txt", "w")) to save the terminal text directly into a permanent file on your hard drive.
* **flush=True (The Immediate Delivery Guy):** Computers usually buffer (hold onto) text in memory chunks before dumping it to the screen or file all at once to save system resources. Setting flush=True forces Python to bypass the waiting room and instantly project the text onto the screen. This is crucial for real-time progress bars or critical system error logging.

# Match and Case

````python
score = 85  # Example score

match score:
    case _ if 100 >= score >= 80: # _ acts as a capture variable, and the if applies the condition (guard)
        print("Grade 4.00")
    case _ if score >= 75:
        print("Grade 3.75")
    case _ if score >= 70:
        print("Grade 3.50")
    case _ if score >= 65:
        print("Grade 3.25")
    case _ if score >= 60:
        print("Grade 3.00")
    case _ if score >= 55:
        print("Grade 2.75")
    case _ if score >= 50:
        print("Grade 2.50")
    case _ if score >= 45:
        print("Grade 2.25")
    case _ if score >= 40:
        print("Grade 2.00")
    case _:  # The '_' acts as the 'else' wildcard for anything else
        print("Grade 0.00")
````

````python
students = [
    "Shihab",
    "Abdullah",
    "Jabir",
    "Jabed"
]

for i in range(len(students)):
  name = students[i]
  print(name, end = " - ")
  match name:
    case "Shihab" | "Jabir":
      print("Shah Paran Hall")
    case "Abdullah":
      print("Dhamali Para")
    case "Jabed":
      print("Borogul")
    case _ : # for anyone who is not in our list
      print("Who?")


print(students)
````

> A NameError is a built-in Python error that happens when we try to use a variable, function, or module name that Python cannot find or recognize.

````python
try:
  x = int(input("What's x?\n"))
except ValueError:
  print("x should be an integer")
else:
  print(f'x is {x}')
````

This code throws a **NameError** because if a user inputs something that is not an integer (like the letter "*a*"), the line x = int(...) *crashes* before it can actually **create the variable x**

An example from [google](https://share.google/aimode/8TvLaXlENOVVyXEvq)

````python
from statistics import mean, variance

# 1. Ask for input (Notice: DO NOT wrap input() in float() here!)
data_input = input("Enter your sample data points (you can use spaces or commas):\n")

# 2. Clean the input: Replace commas with spaces, then split the text into a list
cleaned_input = data_input.replace(",", " ").split()

# 3. Safely convert each split piece of text into a real float number
sample_data = [float(x) for x in cleaned_input]

# 4. Calculate results if you have enough data points
sample_size = len(sample_data)

if sample_size > 1:
    degrees_of_freedom = sample_size - 1
    print(f"\n--- Statistical Results ---")
    print(f"Sample Size (n): {sample_size}")
    print(f"Degrees of Freedom (df): {degrees_of_freedom}")
    print(f"Sample Mean: {mean(sample_data):.2f}")
    print(f"Sample Variance: {variance(sample_data):.2f}")
else:
    print("Error: You must enter at least two numbers to calculate variance.")
````

````python
students = []

# 1. Initialize the empty list before the loop
students = []

with open("sample_data/california_housing_test.csv") as file:
    header = next(file)

    for line in file:
        row = line.rstrip().split(',')

        lng = row[0]
        lat = row[1]

        student = {"longitude" : lng, "latitude": lat}
        students.append(student)

# 2. Fix dictionary lookups using square brackets and string keys
def get_lat(student):
    return float(student["latitude"])

def get_lon(student):
    return float(student["longitude"])

# 3. Slice AFTER sorting to get the true top 5 highest longitudes
# (Slicing before sorting only sorts the first 5 arbitrary rows)
for student in sorted(students, key = get_lon, reverse = True)[:5]:
    print(f"Location is at Long: {student['longitude']}, Lat: {student['latitude']}")
````

````python
# 1. Initialize the empty list before the loop
students = []

with open("sample_data/california_housing_test.csv") as file:
    header = next(file)

    for line in file:
        row = line.rstrip().split(',')

        lng = row[0]
        lat = row[1]

        student = {"longitude" : lng, "latitude": lat}
        students.append(student)

# 2. Fix dictionary lookups using square brackets and string keys
def get_lat(student):
    return float(student["latitude"])

def get_lon(student):
    return float(student["longitude"])

# 3. Slice AFTER sorting to get the true top 5 highest longitudes
# (Slicing before sorting only sorts the first 5 arbitrary rows)
for student in sorted(students, key = lambda student: student['latitude'], reverse = True)[:5]:
    print(f"Location is at Long: {student['longitude']}, Lat: {student['latitude']}")
````

# Math

````python
import math

# Direct constant
print(math.e)  # Output: 2.718281828459045

# Using the exponential function e^1
print(math.exp(1))  # Output: 2.718281828459045
````

````python
import math

def calculate_e_taylor(iterations=20):
    e_approximation = 0
    for n in range(iterations):
        e_approximation += 1 / math.factorial(n)
    return e_approximation

print(calculate_e_taylor(10000))
````

````python
def calculate_e_limit(n=100000000):
    return (1 + 1 / n) ** n

print(calculate_e_limit())  # Output: 2.7182818148460567 (slight rounding error due to float limits)
````
